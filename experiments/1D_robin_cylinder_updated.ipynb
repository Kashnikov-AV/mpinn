```python
# Физические параметры для задачи Робина (цилиндрическая геометрия)
phys = PhysicsParams(
    x_left=0.1, x_right=1.0,
    T_left=500.0, T_inf=300.0,
    _lambda=1.0, h=10.0
)
phys
```

```python
base_config = TrainConfig(
    hidden_features=64, num_layers=2, activation_name='GELU',
    opt_name='adam', lr=0.01, max_epochs=3000, patience=200,
    num_points=100, weights=(1.0, 1.0, 1.0)
)
base_config
```

```python
# Граничные условия: Дирихле слева, Робин справа
bc_fns = [
    partial(dirichlet_bc, x=jnp.array([phys.x_left]), T=phys.T_left),
    partial(robin_bc, x=jnp.array([phys.x_right]), alpha=phys.alpha_right, beta=phys.beta_right, h=phys.gamma_right)
]
```

```python
config = TrainConfig(
    hidden_features=64, num_layers=2, activation_name='GELU',
    opt_name='adam', lr=0.01, max_epochs=3000, patience=200,
    num_points=100, weights=(1.0, 1.0, 1.0),
    save_img=True, show_plot=True, image_path='images/pinn/1D_cylinder_robin.png'
)
metrics, history = run_experiment(
    config=config, phys=phys, pde_fn=cylinder_1d,
    exact_fn=cylinder_1d_robin_exact, bc_fns_override=bc_fns
)
metrics
```

```python
show_history(history, save_path='images/pinn/1D_cylinder_robin_history.png', show_plot=True)
```

```python
param_grid = {
    'hidden_features': [32, 64, 128],
    'num_layers': [1, 2, 3],
    'activation_name': ['tanh', 'Swish', 'GELU', 'ReLU'],
    'num_points': [10, 20, 50, 100],
}
```

```python
df_results = run_grid_search(
    param_grid=param_grid, base_config=base_config, phys=phys,
    csv_path='csv_results/1D_cylinder_robin_results.csv',
    pde_fn=cylinder_1d, exact_fn=cylinder_1d_robin_exact
)
df_results.head()
```
