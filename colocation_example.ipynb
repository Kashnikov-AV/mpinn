```python
import matplotlib.pyplot as plt
from mpinn.geom import Rectangle, Circle, Box, Sphere
import os
import jax.numpy as jnp

os.makedirs('images', exist_ok=True)
plt.rcParams.update({'font.size': 12})

# --- 2D Примеры ---
print("Генерация 2D...")

# Прямоугольник
rect = Rectangle(0, 2, 0, 1)
# generate_collocation возвращает ОДИН массив (N, dim), где сначала внутренние, потом граничные
all_points_r = rect.generate_collocation(n_interior=50, n_boundary=40)

# Разделяем точки вручную: первые n_interior - внутренние, остальные - граничные
n_int_r = 50
int_r = all_points_r[:n_int_r]
bc_r = all_points_r[n_int_r:]

# Круг
circ = Circle(center=(0, 0), radius=1.0)
all_points_c = circ.generate_collocation(n_interior=100, n_boundary=40)
n_int_c = 100
int_c = all_points_c[:n_int_c]
bc_c = all_points_c[n_int_c:]

fig, ax = plt.subplots(1, 2, figsize=(10, 5))

# Передаем разделенные массивы в plot_domain
# Если plot_domain требует словарь, создаем его вручную
rect.plot_domain(int_r, {'boundary': bc_r}, ax=ax[0], title="Rectangle")
circ.plot_domain(int_c, {'boundary': bc_c}, ax=ax[1], title="Circle")

plt.savefig('images/geom_2d.png', dpi=150, bbox_inches='tight')
plt.show()

# --- 3D Примеры ---
print("Генерация 3D...")

# Box
box = Box(0, 1, 0, 1, 0, 1)
all_points_b = box.generate_collocation(n_interior=200, n_boundary=60)
n_int_b = 200
int_b = all_points_b[:n_int_b]
bc_b = all_points_b[n_int_b:]

# Сфера
sph = Sphere(center=(0, 0, 0), radius=1.0)
all_points_s = sph.generate_collocation(n_interior=300, n_boundary=80)
n_int_s = 300
int_s = all_points_s[:n_int_s]
bc_s = all_points_s[n_int_s:]

fig = plt.figure(figsize=(12, 5))
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
ax2 = fig.add_subplot(1, 2, 2, projection='3d')

box.plot_domain(int_b, {'boundary': bc_b}, ax=ax1, title="Box")
sph.plot_domain(int_s, {'boundary': bc_s}, ax=ax2, title="Sphere")

plt.savefig('images/geom_3d.png', dpi=150, bbox_inches='tight')
plt.show()

print("Готово! Графики сохранены в папке images/")
```
